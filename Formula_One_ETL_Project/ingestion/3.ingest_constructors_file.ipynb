#####Step 1 - Read the JSON file using the spark dataframe reader

```python
constructors_schema = "constructorId INT, constructorRef STRING, name STRING, nationality STRING"
```

```python
constructors_df = spark.read \
    .option("header", True) \
    .schema(constructors_schema) \
    .json(f"{raw_folder_path}/constructors.json")
```

#####Step 2 - Drop unwanted columns from the dataframe

```python
from pyspark.sql.functions import col
```

```python
constructors_dropped_df = constructors_df.drop(col('url'))
```

#####Step 3 - Rename columns and add ingestion date

```python
from pyspark.sql.functions import current_timestamp, lit
```

```python
constructors_rename_df = constructors_dropped_df \
    .withColumnRenamed('constructorId', 'constructor_id') \
    .withColumnRenamed('constructorRef', 'constructor_ref') \
    .withColumn("data_source", lit(v_data_source))
```

```python
constructors_final_df = add_ingestion_date(constructors_rename_df)
```

#####Step 4 - Write output to parquet file

```python
constructors_final_df.write.mode("overwrite").parquet(f"{processed_folder_path}/constructors")
```
